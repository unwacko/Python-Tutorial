Mathematical Operators:

```python
print("Addition:", 5 + 3) # Adds two operands
print("Subtraction:", 10 - 4) # Subtracts right operand from left operand
print("Multiplication:", 6 * 7) # Multiplies two operands
print("Division (float):", 15 / 4) # Divides left operand by right operand (float result)
print("Division (integer // floor division):", 15 // 4) # Divides left operand by right operand (integer result, floors the division)
print("Modulus (remainder):", 10 % 3) # Returns the remainder of the division
print("Exponentiation:", 2 ** 3) # Raises the first operand to the power of the second
```

Data Grouping

```python
# --- Lists ---
print("\n--- Python Lists ---")

# 1. Creation
my_list = [10, 20, 30, 40, 50, 'apple', 'banana', 60.5]
print("Initial list:", my_list)

# 2. Accessing Elements (Indexing)
print("First element (index 0):", my_list[0])
print("Last element (negative indexing):", my_list[-1])

# 3. Slicing
print("Slice from index 2 to 5 (exclusive):", my_list[2:6])
print("Slice from beginning to index 3 (exclusive):", my_list[:4])
print("Slice from index 4 to end:", my_list[4:])
print("Slice with a step:", my_list[::2])
print("Reversed list using slicing:", my_list[::-1])

# 4. Adding Elements
my_list.append('cherry') # Add to the end
print("After append 'cherry':", my_list)

my_list.insert(2, 'orange') # Insert at a specific index
print("After insert 'orange' at index 2:", my_list)

my_other_list = [70, 80]
my_list.extend(my_other_list) # Extend with elements from another iterable
print("After extend with [70, 80]:", my_list)

# 5. Modifying Elements
my_list[1] = 25 # Modify an element by index
print("After modifying element at index 1 to 25:", my_list)

# 6. Removing Elements
my_list.remove('banana') # Remove by value (first occurrence)
print("After remove 'banana':", my_list)

popped_element = my_list.pop(0) # Remove by index and return the element
print("After pop from index 0:", my_list, "(Popped:", popped_element, ")")

del my_list[-2] # Delete element by index
print("After del element at index -2:", my_list)

# my_list.clear() # Removes all items from the list
# print("After clear():", my_list)

# 7. List Information
print("Length of the list:", len(my_list))
print("'apple' in list?:", 'apple' in my_list)
print("Index of 'apple':", my_list.index('apple'))
print("Count of 25 (if multiple):", my_list.count(25))

# 8. Concatenation
list_a = [1, 2]
list_b = [3, 4]
list_c = list_a + list_b
print("Concatenated list_a and list_b:", list_c)

# 9. Iteration
print("Iterating through list:")
for item in my_list:
    print(f"  - {item}")


# --- Tuples ---
print("\n--- Python Tuples ---")

# 1. Creation
my_tuple = (1, 2, 3, 'hello', 4.5, (6, 7))
print("Initial tuple:", my_tuple)

# A tuple with a single element needs a trailing comma
single_element_tuple = (1,)
print("Single element tuple:", single_element_tuple)

# 2. Accessing Elements (Indexing)
print("First element:", my_tuple[0])
print("Nested tuple element:", my_tuple[5][0])

# 3. Slicing
print("Slice from index 1 to 4:", my_tuple[1:5])
print("Tuple reversed:", my_tuple[::-1])

# 4. Immutability (Tuples cannot be changed after creation)
# my_tuple[0] = 99 # This would raise a TypeError
print("Tuples are immutable. Cannot modify elements directly.")

# 5. Concatenation (creates a new tuple)
tuple_a = (10, 20)
tuple_b = ('a', 'b')
tuple_c = tuple_a + tuple_b
print("Concatenated tuples:", tuple_c)

# 6. Tuple Information
print("Length of the tuple:", len(my_tuple))
print("'hello' in tuple?:", 'hello' in my_tuple)
print("Index of 'hello':", my_tuple.index('hello'))
print("Count of 2:", my_tuple.count(2))

# 7. Iteration
print("Iterating through tuple:")
for element in my_tuple:
    print(f"  - {element}")

# 8. Tuple Packing and Unpacking
coordinates = (10, 20, 30)
x, y, z = coordinates # Unpacking
print(f"Unpacked coordinates: x={x}, y={y}, z={z}")


# --- Ranges ---
print("\n--- Python Ranges ---")

# 1. Creation
# range(stop)
range_1 = range(5) # 0, 1, 2, 3, 4
print("range(5):")
for i in range_1: print(f"  - {i}")

# range(start, stop)
range_2 = range(2, 7) # 2, 3, 4, 5, 6
print("range(2, 7):")
for i in range_2: print(f"  - {i}")

# range(start, stop, step)
range_3 = range(0, 10, 2) # 0, 2, 4, 6, 8
print("range(0, 10, 2):")
for i in range_3: print(f"  - {i}")

range_4 = range(10, 0, -1) # 10, 9, ..., 1
print("range(10, 0, -1):")
for i in range_4: print(f"  - {i}")

# 2. Convert to list (to view all elements)
print("range(5) as a list:", list(range(5)))

# 3. Accessing elements (like a list/tuple, Python 3.x ranges support this)
print("Element at index 2 in range(0, 10, 2):", range_3[2])

# 4. Slicing (creates a new range object)
sub_range = range_3[1:4]
print("Sliced range_3[1:4] as a list:", list(sub_range))

# 5. Check presence
print("4 in range(0, 10, 2)?:", 4 in range_3)
print("5 in range(0, 10, 2)?:", 5 in range_3)

# 6. Length
print("Length of range(0, 10, 2):", len(range_3))


# --- Dictionaries ---
print("\n--- Python Dictionaries ---")

# 1. Creation
my_dict = {
    'name': 'Alice',
    'age': 30,
    'city': 'New York',
    'occupations': ['Engineer', 'Developer'],
    1: 'one'
}
print("Initial dictionary:", my_dict)

# 2. Accessing Values
print("Name:", my_dict['name'])
print("Age (using .get() - safer):", my_dict.get('age'))
print("Occupation at index 0:", my_dict['occupations'][0])
# print(my_dict['country']) # This would raise a KeyError
print("Non-existent key using .get() (returns None):", my_dict.get('country'))
print("Non-existent key using .get() with default value:", my_dict.get('country', 'Unknown'))

# 3. Adding/Modifying Key-Value Pairs
my_dict['email'] = 'alice@example.com' # Add new key-value pair
print("After adding email:", my_dict)

my_dict['age'] = 31 # Modify existing value
print("After modifying age:", my_dict)

# 4. Removing Key-Value Pairs
removed_city = my_dict.pop('city') # Remove by key and return value
print("After pop 'city':", my_dict, "(Removed:", removed_city, ")")

del my_dict['occupations'] # Delete by key
print("After del 'occupations':", my_dict)

# 5. Views of Keys, Values, Items
print("Dictionary keys:", my_dict.keys())
print("Dictionary values:", my_dict.values())
print("Dictionary items:", my_dict.items())

# 6. Length
print("Length of dictionary:", len(my_dict))

# 7. Check presence (for keys)
print("'name' in dictionary?:", 'name' in my_dict)
print("'Alice' in dictionary (checks keys, not values)?:", 'Alice' in my_dict)
print("'Alice' in values?:", 'Alice' in my_dict.values())

# 8. Iteration
print("Iterating through dictionary keys:")
for key in my_dict:
    print(f"  Key: {key}, Value: {my_dict[key]}")

print("Iterating through dictionary items:")
for key, value in my_dict.items():
    print(f"  Key: {key}, Value: {value}")

# 9. Update method
another_dict = {'gender': 'Female', 'age': 32} # 'age' will be updated
my_dict.update(another_dict)
print("After update with another_dict:", my_dict)


# --- Arrays (using array module) ---
print("\n--- Python Arrays (using 'array' module) ---")
import array

# 1. Creation
# 'i' is the type code for signed integer (typically 2 or 4 bytes)
my_array = array.array('i', [10, 20, 30, 40, 50])
print("Initial array:", my_array)

# Type codes reference:
# 'b': signed char (1 byte)
# 'B': unsigned char (1 byte)
# 'h': signed short (2 bytes)
# 'H': unsigned short (2 bytes)
# 'i': signed int (2 or 4 bytes)
# 'I': unsigned int (2 or 4 bytes)
# 'l': signed long (4 bytes)
# 'L': unsigned long (4 bytes)
# 'q': signed long long (8 bytes)
# 'Q': unsigned long long (8 bytes)
# 'f': float (4 bytes)
# 'd': double (8 bytes)

float_array = array.array('f', [1.1, 2.2, 3.3])
print("Float array:", float_array)

# 2. Accessing Elements (Indexing)
print("First element:", my_array[0])
print("Last element:", my_array[-1])

# 3. Slicing
print("Slice from index 1 to 3:", my_array[1:4])

# 4. Adding Elements
my_array.append(60) # Add to the end
print("After append 60:", my_array)

my_array.insert(2, 25) # Insert at a specific index
print("After insert 25 at index 2:", my_array)

my_other_array = array.array('i', [70, 80])
my_array.extend(my_other_array) # Extend with elements from another array
print("After extend with another array:", my_array)

# 5. Modifying Elements
my_array[1] = 22 # Modify an element by index
print("After modifying element at index 1 to 22:", my_array)

# 6. Removing Elements
removed_val = my_array.pop() # Remove last element and return it
print("After pop (last element removed):", my_array, "(Removed:", removed_val, ")")

my_array.remove(30) # Remove first occurrence of value
print("After remove 30:", my_array)

# 7. Array Information
print("Length of the array:", len(my_array))
print("Index of 40:", my_array.index(40))
print("Count of 20:", my_array.count(20))

# 8. Concatenation (creates a new array)
array_a = array.array('i', [1, 2])
array_b = array.array('i', [3, 4])
array_c = array_a + array_b
print("Concatenated arrays:", array_c)

# 9. Iteration
print("Iterating through array:")
for num in my_array:
    print(f"  - {num}")

# Important note: For more advanced array operations, especially with numerical data,
# the 'numpy' library is highly recommended for its performance and extensive functionalities.
# import numpy as np
# numpy_array = np.array([1, 2, 3, 4, 5])
# print("\n--- NumPy Array Example ---")
# print(numpy_array)
```

Loops

```python
# --- If/Elif/Else Statements ---
print("\n--- If/Elif/Else Statements ---")

age = 25

if age < 13:
    print("You are a child.")
elif age < 18:
    print("You are a teenager.")
elif age < 65:
    print("You are an adult.")
else:
    print("You are a senior citizen.")

# Nested If-Else
is_sunny = True
temperature = 30

if is_sunny:
    if temperature > 25:
        print("It's a hot and sunny day!")
    else:
        print("It's a pleasant sunny day.")
else:
    print("It's not sunny.")

# Short-hand If (Ternary Operator)
a = 10
b = 20
print("A is greater" if a > b else "B is greater or equal")


# --- For Loops ---
print("\n--- For Loops ---")

# Basic for loop with a list
fruits = ["apple", "banana", "cherry"]
print("Iterating through a list:")
for fruit in fruits:
    print(f"  - {fruit}")

# For loop with range()
print("Counting from 0 to 4:")
for i in range(5):
    print(f"  - {i}")

print("Counting from 2 to 6:")
for i in range(2, 7):
    print(f"  - {i}")

print("Counting by 2s from 0 to 8:")
for i in range(0, 10, 2):
    print(f"  - {i}")

# For loop with string
word = "Python"
print("Iterating through a string:")
for char in word:
    print(f"  - {char}")

# For loop with a dictionary
person = {"name": "John", "age": 30, "city": "New York"}
print("Iterating through dictionary keys:")
for key in person:
    print(f"  Key: {key}, Value: {person[key]}")

print("Iterating through dictionary items (key-value pairs):")
for key, value in person.items():
    print(f"  Key: {key}, Value: {value}")

# For loop with break statement
print("For loop with break:")
for i in range(10):
    if i == 5:
        print("  Breaking loop at 5")
        break
    print(f"  - {i}")

# For loop with continue statement
print("For loop with continue:")
for i in range(5):
    if i == 2:
        print("  Skipping 2")
        continue
    print(f"  - {i}")

# For loop with else block (executes if loop completes without break)
print("For loop with else block:")
for i in range(3):
    print(f"  - {i}")
else:
    print("  Loop finished without breaking.")

print("For loop with else block (demonstrating break preventing else):")
for i in range(5):
    if i == 2:
        print("  Breaking loop at 2, else will not execute.")
        break
    print(f"  - {i}")
else:
    print("  This message will NOT print.")

# Enumerate for index and value
print("Iterating with enumerate (index and value):")
for index, item in enumerate(fruits):
    print(f"  Index {index}: {item}")

# Zip for parallel iteration
list1 = [1, 2, 3]
list2 = ['a', 'b', 'c']
print("Iterating with zip (parallel iteration):")
for num, char in zip(list1, list2):
    print(f"  Number: {num}, Character: {char}")
```
